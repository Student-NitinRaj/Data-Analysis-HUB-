```python
category_sales = df.groupby("Category")["Total_Sales_USD"].sum()

plt.figure(figsize=(8,5))

category_sales.plot(kind='bar')

plt.title("Total Sales by Category")

plt.xlabel("Category")

plt.ylabel("Sales USD")

plt.show()    # Kaunsi category sabse jyada sales generate kar rahi hai.
```

```python
city_sales = df.groupby("Customer_City")["Total_Sales_USD"].sum()

plt.figure(figsize=(8,5))

city_sales.plot(kind='bar')

plt.title("City Wise Sales")

plt.xlabel("City")

plt.ylabel("Total Sales")

plt.show()      # Kaunsi city me highest revenue aa raha hai.
```

```python
top_products = df.groupby("Product_Name")["Quantity_Sold"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))

top_products.plot(kind='bar')

plt.title("Top Selling Products")

plt.xlabel("Product")

plt.ylabel("Quantity Sold")

plt.show()
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

df["Month"] = df["Order_Date"].dt.month
```

```python
monthly_sales = df.groupby("Month")["Total_Sales_USD"].sum()

plt.figure(figsize=(10,5))

monthly_sales.plot(marker='o')

plt.title("Monthly Sales Trend")

plt.xlabel("Month")

plt.ylabel("Sales")

plt.grid()

plt.show()      # Kis month me sales peak hui.
```

```python
plt.figure(figsize=(8,5))

sns.heatmap(df.corr(numeric_only=True), annot=True)

plt.title("Correlation Heatmap")

plt.show()      # Price aur Sales ka relation pata chalega.
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Total_Sales_USD"], bins=20)

plt.title("Sales Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(x="Price_USD", y="Total_Sales_USD", data=df)

plt.title("Price vs Sales")

plt.show()
```
