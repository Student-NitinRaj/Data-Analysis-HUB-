```python
plt.figure(figsize=(8,5))
ax= sns.countplot(data=df,x="Gender")
ax.bar_label(ax.containers[0])
plt.title("Gender Basic Sales")
plt.show()
```

From the above analysis and visullization show Gender basis minor diff of Sales but Female are higer buyer APPLE PRODUCT

```python
# Which State Higher Sales On IPHONE
plt.figure(figsize=(6,4))
ax= sns.countplot(data=df,x="State")
ax.bar_label(ax.containers[0])
plt.title("State Basic Sales")
plt.show()
```

From Above Vissulization show Maharashtra and Karnataka has higer sales on Apple Product

```python
# Visualization on the basis of Payment Mode
plt.figure(figsize=(6,4))
ax=sns.countplot(data=df,x="Payment Mode")
ax.bar_label(ax.containers[0])
plt.title("Payment Mode Basic Sales")
plt.show()
```

From above Visulization show via upi higer payment Mode

```python
# Which Product are higher sales
plt.figure(figsize=(10,4))
ax= sns.countplot(data=df,x="Product")
ax.bar_label(ax.containers[0])
plt.title("Product Basic Sales")
plt.show()
```

```python
# Which Product are higher sales
plt.figure(figsize=(8, 8))
product_counts = df['Product'].value_counts()
plt.pie(product_counts, labels=product_counts.index, autopct='%1.1f%%', startangle=140)
plt.title("Product Basic Sales")
plt.show()
```

From Above Visulization show all product are same sales but iphone 15 Pro, MacBook Pro, AirPods Pro on higer sales

```python
# Which Product are higher sales
plt.figure(figsize=(8, 8))
product_counts = df['Profit'].value_counts()
plt.pie(product_counts, labels=product_counts.index, autopct='%1.1f%%', startangle=140)
plt.title("Profit on the Persentage")
plt.show()
```

```python
gb=df.groupby("Product")["Profit"].sum()
gb
```

```python
df.sample(3)
```

```python
gb1=df.groupby("Product").agg({"Category":"count","Profit":"mean","Amount":"mean"})
gb1
```

```python
sns.heatmap(gb1, annot=True, cmap="Reds")
plt.show()
```

```python
gb2=df.groupby("City").agg({"Category":"count","Profit":"mean","Amount":"mean"})
gb2
```

```python
sns.heatmap(gb2, annot=True, cmap="Greens")
plt.show()
```

From Above data analysis show on Chennai, bengaluru, Coimbatore higher profit

```python
gb3=df.groupby("Gender").agg({"Product Price":"sum","Amount":"sum","Profit":"sum"})
gb3
```

```python
sns.heatmap(gb3,annot=True,cmap="Blues")
plt.show()
```

From above Visulation female has higer buyer of Apple Product
